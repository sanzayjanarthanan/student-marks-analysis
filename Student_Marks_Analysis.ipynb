```python
students = {"Arun": 78, "Priya": 92, "Karthik": 65,
            "Divya": 88, "Rahul": 54, "Meena": 71}

marks = list(students.values())

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

top = max(students, key=students.get)
bottom = min(students, key=students.get)

print(f"Total   : {total}")
print(f"Average : {average:.2f}")
print(f"Highest : {highest} ({top})")
print(f"Lowest  : {lowest} ({bottom})")
```
